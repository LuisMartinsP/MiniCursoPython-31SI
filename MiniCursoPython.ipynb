###Exercício 8: Classe Produto

Implemente uma classe Produto, que terá: nome, preço e quantidade.

Defina métodos: Construtor (_ _init_ _), Imprimir( _ _str_ _ )

Funções:

comprar(qtd) -> aumenta quantidade do produto em estoque

vender(qtd) -> diminui quantidade do produto em estoque

Verificações: produto não deve poder ter quantidade negativa, se atentar a isso.


```python
class Produto:
  def __init__(self, nome, preco, qtd):
    self.nome = nome
    self.preco = preco
    self.qtd = qtd

  def __str__(self):
    return(f"Nome : {self.nome}\nPreco: {self.preco} \nQuantidade: {self.qtd}")

  def comprar(self, qtd_comprada):
    self.qtd += qtd_comprada

  def vender(self, qtd_vendida):
    if self.qtd - qtd_vendida < 1:
      self.qtd = 0
      print("Acabou o produto")
    else:
      self.qtd -= qtd_vendida

arroz = Produto("Arroz", "1.90", 10)
print(arroz)
```

```python
arroz.comprar(5)
print(arroz)
```

```python
arroz.vender(10)
print(arroz)
```

```python
import os
import requests


def baixar_sprites(pokemon):

    nome = pokemon.nome.lower()

    os.makedirs("sprites", exist_ok=True)

    url_frente = f"https://play.pokemonshowdown.com/sprites/ani/{nome}.gif"
    url_costas = f"https://play.pokemonshowdown.com/sprites/ani-back/{nome}.gif"

    caminho_frente = f"sprites/{nome}_front.gif"
    caminho_costas = f"sprites/{nome}_back.gif"

    resposta_frente = requests.get(url_frente)
    if resposta_frente.status_code == 200:
        with open(caminho_frente, "wb") as arquivo:
            arquivo.write(resposta_frente.content)
        pokemon.imagem_frente = caminho_frente
    else:
        print(f"Sprite de frente não encontrado para {pokemon.nome}")

    resposta_costas = requests.get(url_costas)
    if resposta_costas.status_code == 200:
        with open(caminho_costas, "wb") as arquivo:
            arquivo.write(resposta_costas.content)
        pokemon.imagem_costas = caminho_costas
    else:
        print(f"Sprite de costas não encontrado para {pokemon.nome}")
```

```python
import gradio as gr
import random
import os
import base64


def _musica(batalha):
    if batalha.acabou():
        return "musicas/victory.mp3"

    return "musicas/battle.mp3"


def _sprite_base64(caminho):
    if caminho is None:
        return ""
    with open(caminho, "rb") as f:
        dados = base64.b64encode(f.read()).decode()
    return f"data:image/gif;base64,{dados}"


def _arena_html(batalha):
    c1 = batalha.combatente1
    c2 = batalha.combatente2

    back = _sprite_base64(c1.pokemon.imagem_costas)
    front = _sprite_base64(c2.pokemon.imagem_frente)

    if not back or not front:
        return ""

    return f"""
    <div class="arena">
        <div class="plataforma plataforma-oponente"></div>
        <div class="plataforma plataforma-jogador"></div>
        <img class="sprite-arena sprite-oponente" src="{front}">
        <img class="sprite-arena sprite-jogador" src="{back}">
    </div>
    """


def _texto_info(combatente):
    p = combatente.pokemon
    total = 20
    cheios = int((combatente.hp_atual / p.hp) * total)
    cheios = max(0, cheios)
    barra = "█" * cheios + "░" * (total - cheios)

    linhas = [
        f"{p.nome}",
        f"HP {barra}",
        f"{combatente.hp_atual}/{p.hp}",
    ]
    if combatente.esta_carregando:
        linhas.append(
            f"⏳ {combatente.ataque_carregando['nome']} "
            f"({combatente.turnos_restantes}t)"
        )
    return "\n".join(linhas)


def _estado_ui(batalha):
    c1, c2 = batalha.combatente1, batalha.combatente2
    acabou = batalha.acabou()

    dd1 = gr.update(
        choices=[a["nome"] for a in c1.ataques_especiais],
        value=None,
        interactive=not acabou and not c1.esta_carregando,
    )
    dd2 = gr.update(
        choices=[a["nome"] for a in c2.ataques_especiais],
        value=None,
        interactive=not acabou and not c2.esta_carregando,
    )

    if acabou:
        status = f"Vencedor: {batalha.vencedor().nome}"
    else:
        status = "Escolham os ataques do próximo turno."

    return (
        _arena_html(batalha),
        _texto_info(c1),
        _texto_info(c2),
        status,
        dd1,
        dd2,
        batalha.log_texto(),
    )


def iniciar_batalha(nome1, nome2):
    pokemon1 = next(p for p in pokedex if p.nome == nome1)
    pokemon2 = next(p for p in pokedex if p.nome == nome2)

    batalha = Batalha(Combatente(pokemon1), Combatente(pokemon2))
    batalha._registrar("A batalha começou!")

    return (batalha, _musica(batalha), *_estado_ui(batalha))


def executar_turno(batalha, ataque1, ataque2):
    if batalha is None:
        return (batalha, gr.update(), "", "", "",
                "Inicie uma batalha primeiro.", gr.update(), gr.update(), "")

    if batalha.acabou():
        return (batalha, _musica(batalha), *_estado_ui(batalha))

    batalha.executar_turno(ataque1, ataque2)
    return (batalha, _musica(batalha), *_estado_ui(batalha))


custom_css = """
@import url('https://fonts.googleapis.com/css2?family=Press+Start+2P&display=swap');

.gradio-container {
    font-family: 'Press Start 2P', monospace !important;
    background: linear-gradient(135deg, #1a1a2e 0%, #16213e 100%) !important;
}

.gradio-container .audio-container {
    max-height: 0px;
    margin: 0 auto;
    opacity: 0.6;
}

.gradio-container h1 {
    text-align: center;
    color: #ffcb05 !important;
    text-shadow: 3px 3px 0 #3b4cca, 6px 6px 0 #000;
    font-size: 2em !important;
    padding: 20px;
}

.gradio-container label,
.gradio-container .label-wrap span,
.gradio-container textarea,
.gradio-container input {
    font-family: 'Press Start 2P', monospace !important;
    font-size: 0.7em !important;
}

.gradio-container button {
    font-family: 'Press Start 2P', monospace !important;
    background: #ee1515 !important;
    color: white !important;
    border: 3px solid #000 !important;
    border-radius: 8px !important;
    box-shadow: 4px 4px 0 #000 !important;
    transition: all 0.1s !important;
}

.gradio-container button:hover {
    transform: translate(2px, 2px);
    box-shadow: 2px 2px 0 #000 !important;
}

.pokemon-card {
    background: rgba(255, 255, 255, 0.95) !important;
    border: 4px solid #000 !important;
    border-radius: 12px !important;
    padding: 15px !important;
    box-shadow: 6px 6px 0 #000 !important;
}

.log-box textarea {
    background: #000 !important;
    color: #0f0 !important;
    font-family: 'Courier New', monospace !important;
    border: 2px solid #0f0 !important;
}

/* === ARENA === */

.arena {
    position: relative;
    width: 100%;
    height: 340px;
    background: linear-gradient(180deg,
        #87ceeb 0%,
        #a8dcf0 40%,
        #7ec850 45%,
        #4a9c3f 100%);
    border: 4px solid #000;
    border-radius: 12px;
    box-shadow: 6px 6px 0 #000;
    overflow: hidden;
    margin-bottom: 15px;
}

.plataforma {
    position: absolute;
    border-radius: 50%;
    background: radial-gradient(ellipse at center,
        rgba(80, 50, 20, 0.55) 0%,
        rgba(80, 50, 20, 0) 70%);
}

.plataforma-jogador {
    bottom: 25px;
    left: 8%;
    width: 260px;
    height: 70px;
}

.plataforma-oponente {
    top: 100px;
    right: 10%;
    width: 180px;
    height: 45px;
}

.sprite-arena {
    position: absolute;
    image-rendering: pixelated;
    filter: drop-shadow(3px 3px 0 rgba(0,0,0,0.4));
}

.sprite-jogador {
    bottom: 45px;
    left: 14%;
    height: 150px;
}

.sprite-oponente {
    top: 75px;
    right: 16%;
    height: 110px;
}
@keyframes shake-atacar {
    0%, 100% { transform: translateX(0); }
    50% { transform: translateX(20px) scale(1.1); }
}

@keyframes shake-atacar-inverso {
    0%, 100% { transform: translateX(0); }
    50% { transform: translateX(-20px) scale(1.1); }
}

@keyframes piscar-dano {
    0%, 100% { opacity: 1; filter: none; }
    25%, 75% { opacity: 0.3; filter: brightness(3) saturate(0); }
}

/* jogador ataca no primeiro momento (0 a 0.5s) */
.sprite-jogador.ataca-primeiro { animation: shake-atacar 0.5s ease-out; }
.sprite-oponente.apanha-primeiro { animation: piscar-dano 0.5s ease-in-out; }

/* oponente ataca no primeiro momento */
.sprite-oponente.ataca-primeiro { animation: shake-atacar-inverso 0.5s ease-out; }
.sprite-jogador.apanha-primeiro { animation: piscar-dano 0.5s ease-in-out; }

/* segundo momento (com delay de 1s) */
.sprite-jogador.ataca-segundo {
    animation: shake-atacar 0.5s ease-out 1s;
}
.sprite-oponente.apanha-segundo {
    animation: piscar-dano 0.5s ease-in-out 1s;
}
.sprite-oponente.ataca-segundo {
    animation: shake-atacar-inverso 0.5s ease-out 1s;
}
.sprite-jogador.apanha-segundo {
    animation: piscar-dano 0.5s ease-in-out 1s;
}
"""

nomes = [pokemon.nome for pokemon in pokedex]

with gr.Blocks(theme=gr.themes.Soft(primary_hue="red"), css=custom_css) as app:

    batalha_state = gr.State(None)

    gr.Markdown("# ⚔️ Pokémon Battle ⚔️")

    musica = gr.Audio(
        autoplay=True,
        visible=True,
        show_label=False,
        loop=True,
        elem_classes="audio-container",
    )

    app.load(
        lambda: "musicas/opening.mp3",
        inputs=None,
        outputs=musica,
    )

    with gr.Row():
        pokemon1 = gr.Dropdown(nomes, label="🔴 Pokémon 1")
        pokemon2 = gr.Dropdown(nomes, label="🔵 Pokémon 2")

    iniciar = gr.Button("▶ Iniciar batalha", size="lg")

    arena = gr.HTML()

    with gr.Row():

        with gr.Column(elem_classes="pokemon-card"):
            info1 = gr.Textbox(label="Pokémon 1", lines=4, interactive=False)
            ataque1 = gr.Dropdown(label="⚡ Ataque")

        with gr.Column(elem_classes="pokemon-card"):
            info2 = gr.Textbox(label="Pokémon 2", lines=4, interactive=False)
            ataque2 = gr.Dropdown(label="⚡ Ataque")

    turno = gr.Textbox(label="📢 Status", interactive=False)

    botao = gr.Button("⚔ Executar turno", size="lg")

    log = gr.Textbox(
        lines=5,
        label="📜 Descrição do Turno:",
        elem_classes="log-box",
        interactive=False,
    )

    iniciar.click(
        iniciar_batalha,
        inputs=[pokemon1, pokemon2],
        outputs=[batalha_state, musica, arena, info1, info2,
                 turno, ataque1, ataque2, log],
    )

    botao.click(
        executar_turno,
        inputs=[batalha_state, ataque1, ataque2],
        outputs=[batalha_state, musica, arena, info1, info2,
                 turno, ataque1, ataque2, log],
    )

app.launch(
    debug=True,
    inline=False,
    share=True,
    allowed_paths=[os.path.abspath("sprites")],
)
```
